```python
import numpy as np
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')
```

1.	Load the patient dataset and show summary with info().

```python
df=pd.read_csv('/content/drive/MyDrive/Datasets/Patient_Data.csv')
df1=df.copy()
df1
```

```python
df1.info()
```

2.	Select only the columns relevant for billing: ['PatientID', 'Department', 'Doctor', 'BillAmount'].

```python
df1[['PatientID','Department','Doctor','BillAmount']]
```

3.	Drop administrative columns like ['ReceptionistID', 'CheckInTime'].

```python
df1 = df1.drop(['ReceptionistID','CheckInTime'],axis=1)
```

```python
df1
```

4.	Use groupby to find total bill amount per department.

```python

department_bill_amount = df1.groupby('Department')['BillAmount'].sum().reset_index()

display(department_bill_amount)
```

5.	Remove duplicate patient records based on PatientID.

```python
df1_new=df1.drop_duplicates().copy()
df1_new
```

6.	Fill missing BillAmount values with the mean bill amount.

```python
df1_new['BillAmount']=df1_new['BillAmount'].fillna(df1_new['BillAmount'].mean())
df1_new
```

7.	Merge the billing dataset with patient dataset on PatientID.

```python
df2=pd.read_csv('/content/drive/MyDrive/Datasets/Billing_Data.csv')
```

```python
merged_df = pd.merge(df1_new, df2, on='PatientID', how='inner')
display(merged_df)
```

8.	Concatenate an additional DataFrame that contains new patients for the current week (row-wise).

```python
new_patients_data = {
    'PatientID': [106, 107],
    'Name': ['Licy', 'George'],
    'Department': ['Pediatrics', 'Neurology'],
    'Doctor': ['Dr. Ann', 'Dr. Alan'],
    'BillAmount': [4500.00, 8000.00],

}
new_patients_df = pd.DataFrame(new_patients_data)
display(new_patients_df)
```

```python
combined_df = pd.concat([df1_new, new_patients_df], ignore_index=True)
display(combined_df)
```

9.	Concatenate new billing category columns like ['InsuranceCovered', 'FinalAmount'] (column-wise).

```python
Final_df = pd.merge(combined_df, df2[['PatientID', 'InsuranceCovered', 'FinalAmount']], on='PatientID', how='left')
display(combined_df)
```
